```python
import numpy as np

def predict_life_expectancy():

    def get_input(prompt, type_func, min_val=None, max_val=None, allow_q=True, extra_check=None, warning_range=None):
        while True:
            val = input(prompt)
            if allow_q and val.lower() == 'q':
                print("Exiting prediction.")
                return 'q'
            try:
                val = type_func(val)
                if min_val is not None and val < min_val:
                    print(f'Value below minimum ({min_val}).')
                elif max_val is not None and val > max_val:
                    print(f'Value above maximum ({max_val}).')
                elif extra_check and not extra_check(val):
                    continue
                else:
                    if warning_range and (val < warning_range[0] or val > warning_range[1]):
                        print('WARNING: value lies outside of training data range and prediction may lose accuracy. Consider rebalancing model.')
                    return val
            except:
                print(f'Please enter a valid {type_func.__name__}.')

    while True:
        full_model_y_n = ''
        while full_model_y_n not in ['y', 'n']:
            full_model_y_n = input('Do you want to use data drawn from medical records? (y/n, or q to quit): ').lower()
            if full_model_y_n == 'q':
                print("Exiting prediction.")
                return
            if full_model_y_n not in ['y', 'n']:
                print('Please input "y" for "Yes" or "n" for "No".')

        if full_model_y_n == 'y':
            year = get_input('What is the year? (or q to quit): ', int, warning_range=(2000, 2015))
            if year == 'q': return
            u5 = get_input('What is the number of deaths under 5 years old per 1000 people? (or q to quit): ', float, warning_range=(2.3, 224.9))
            if u5 == 'q': return
            am = get_input('What is the adult mortality rate? (or q to quit): ', float, warning_range=(49.384, 719.3605))
            if am == 'q': return
            ac = get_input('What is the yearly alcohol consumption in litres per capita? (or q to quit): ', float, warning_range=(0, 17.87))
            if ac == 'q': return
            bmi = get_input('What is the population\'s average BMI? (or q to quit): ', float, warning_range=(19.8, 32.1))
            if bmi == 'q': return
            p = get_input('What is the percentage of polio vaccination coverage? (or q to quit): ', int, warning_range=(8, 99))
            if p == 'q': return
            hb = get_input('What is the percentage of hepatitis B vaccination coverage? (or q to quit): ', int, warning_range=(8, 99))
            if hb == 'q': return
            thin = get_input('What is the percentage of medically recorded thinness among ages 10-19? (or q to quit): ', float, warning_range=(0.1, 27.7))
            if thin == 'q': return
            hiv = get_input('How many HIV cases are there per 1000 live births? (or q to quit): ', float, warning_range=(0.01, 21.68))
            if hiv == 'q': return
            ecd = get_input('Is the economy status developed? (0 for No, 1 for Yes, or q to quit): ', int, extra_check=lambda x: x in [0, 1])
            if ecd == 'q': return
            sch = get_input('What is the average years of schooling for the population? (or q to quit): ', float, warning_range=(1.1, 14.1))
            if sch == 'q': return
            GDP = get_input('What is the GDP per capita? (or q to quit): ', float, warning_range=(148, 112418))
            if GDP == 'q': return
            gdp_l = np.log(GDP)

            sc_u5 = (u5 - 2.3)/(224.9 - 2.3)
            sc_am = (am - 49.384)/(719.3605 - 49.384)
            sc_bmi = (bmi - 19.8)/(32.1-19.8)
            sc_hb = (hb - 12)/(99-12)
            sc_p = (p - 8)/(99-8)
            sc_hiv = (hiv - 0.01)/(21.68-0.01)
            sc_thin = (thin - 0.1)/(27.7-0.1)
            sc_gdp_l = (gdp_l - 4.997212)/(11.629979-4.997212)
            sc_ecd = ecd
            sc_year = (year - 2000)/(2015-2000)
            sc_ac = (ac - 0)/(17.87-0)
            sc_sch = (sch - 1.1)/(14.1-1.1)

            return "Predicted life expectancy: " + str(round((76.3813 + sc_year*0.4487 + sc_u5*-16.8442 + sc_am*-30.8633 +
                    sc_ac*0.8901 + sc_bmi*-2.4692 + sc_p*1.1625 + sc_hb*-0.8044 +
                    sc_thin*-1.3312 + sc_hiv*1.2507 + sc_ecd*0.8520 +
                    sc_sch*0.9995 + sc_gdp_l*3.9510),2))

        if full_model_y_n == 'n':
            year = get_input('What is the year? (or q to quit): ', int, warning_range=(2000, 2015))
            if year == 'q': return
            u5 = get_input('What is the number of deaths under 5 years old per 1000 people? (or q to quit): ', float, warning_range=(2.3, 224.9))
            if u5 == 'q': return
            am = get_input('What is the adult mortality rate? (or q to quit): ', float, warning_range=(49.384, 719.3605))
            if am == 'q': return
            ac = get_input('What is the yearly alcohol consumption in litres per capita? (or q to quit): ', float, warning_range=(0, 17.87))
            if ac == 'q': return
            ecd = get_input('Is the economy status developed? (0 for No, 1 for Yes, or q to quit): ', int, extra_check=lambda x: x in [0, 1])
            if ecd == 'q': return
            sch = get_input('What is the average years of schooling for the population? (or q to quit): ', float, warning_range=(1.1, 14.1))
            if sch == 'q': return
            GDP = get_input('What is the GDP per capita? (or q to quit): ', float, warning_range=(148, 112418))
            if GDP == 'q': return
            gdp_l = np.log(GDP)

            sc_u5 = (u5 - 2.3)/(224.9 - 2.3)
            sc_am = (am - 49.384)/(719.3605 - 49.384)
            sc_gdp_l = (gdp_l - 4.997212)/(11.629979-4.997212)
            sc_ecd = ecd
            sc_year = (year - 2000)/(2015-2000)
            sc_ac = (ac - 0)/(17.87-0)
            sc_sch = (sch - 1.1)/(14.1-1.1)

            return "Predicted life expectancy: " + str(round((75.6883 + sc_year*0.3419 + sc_u5*-16.9706 + sc_am*-29.9728 +
                    sc_ac*1.0791 + sc_ecd*1.2061 + sc_sch*0.6785 + sc_gdp_l*3.4130),2))
```
